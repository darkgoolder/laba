```python
pipe = L.load_pipe()
clip = M.load_clip()
# Загрузи модель и CLIP
"""
pipe - модель генерации изображений
clip - модель и обработчик для измерения сходства изображения с текстом
"""
```

```python
N = 21
c, k = (N - 1) // 6, (N - 1) % 6

a_cfg, b_cfg = M.settings(M.FACTORS[k], c)

print("Режим A:", a_cfg)
print("Режим B:", b_cfg)
# Проверим режими параметров А и В
```

```python
res = M.variant_summary(pipe, clip, N)

print("Вариант:", res["variant"])
print("Фактор:", res["factor"])

for m, a, s, b, d, ratio, sig in res["rows"]:
    print(
        f"{m:9s} "
        f"A={a:.4f} "
        f"s={s:.4f} "
        f"B={b:.4f} "
        f"Δ={d:+.4f} "
        f"|Δ|/s={ratio:.2f} "
        f"|Δ| > 2s={'да' if sig else 'нет'}"
    )
    # Расчитаем метрики для 5 seed
"""
CLIP - насколько изображение соответствует английскому брифу в представлении CLIP;
энергия границ - средняя величина градиента яркости;
контраст - стандартное отклонение яркости;
время - длительность генерации в секундах.
"""
```

```python
from PIL import Image

N = 21
c, k = (N - 1) // 6, (N - 1) % 6
a_cfg, b_cfg = M.settings(M.FACTORS[k], c)

pair = [
    L.generate(
        pipe,
        cfg["prompt"],
        101,
        steps=cfg["steps"],
        guidance=cfg["guidance"],
        size=cfg["size"],
    )[0].resize((384, 384))
    for cfg in (a_cfg, b_cfg)
]

sheet = Image.new("RGB", (768, 384))
sheet.paste(pair[0], (0, 0))
sheet.paste(pair[1], (384, 0))
sheet.save("comparison.png")

sheet
# Создадим сравнительный лист изображений comparison.png
```
